This notebook only **downloads the official CDC files and builds the analytic cohort**. All modelling then runs from the repository.

It downloads NHANES 1999–2018 (10 survey cycles: demographics, blood pressure, cholesterol, HDL, HbA1c, kidney function, body measures, smoking, diabetes and medical-history questionnaires) and the NCHS public-use **Linked Mortality Files** (National Death Index follow-up to 31 December 2019). That is about 120 files, roughly 150 MB.

```python
import shutil, zipfile, pathlib
out = pathlib.Path("/content/nhanes_cohort_bundle.zip")
with zipfile.ZipFile(out, "w", zipfile.ZIP_DEFLATED) as z:
    for p in list(pathlib.Path("data").glob("*.csv*")) + list(pathlib.Path("results").glob("*")):
        z.write(p)
print(out, round(out.stat().st_size / 1e6, 1), "MB")
from google.colab import files
files.download(str(out))
```
